 What is Object-Oriented Programming (OOP)? 
 - OOP is a way of writing code where we group things into "objects", just like how we see things in the real world.

For example, think of a car. A car has:

Data like color, model, speed

And actions like start(), stop(), drive()

In OOP, we create something called a class (like a blueprint), and then we make objects from it (actual cars based on that blueprint).

OOP helps us do 4 main things:

Encapsulation – We keep everything about one thing (like a car) together in one place.

Inheritance – One object can take stuff from another. For example, an "ElectricCar" can inherit things from a regular "Car".

Polymorphism – One action can behave differently depending on the object. For example, the drive() function could behave differently for a car, bike, or airplane.

Abstraction – We hide the complicated stuff and show only what’s needed. Just like you press a button to start a car without knowing how the engine really works.

In short: OOP helps us organize code better, make it reusable, and closer to how we think in real life.

What is a class in OOP?
 - A class is like a blueprint or design for creating objects.
It’s a model that tells Python what an object should look like and what it should be able to do.

Think about a blueprint of a house:

The blueprint defines how the house will look — how many rooms, windows, doors.

But the blueprint itself is not a real house.

You can use it to build many real houses based on the same design.

Similarly, in OOP:

A class defines the properties (like color, size) and behaviors (like drive, fly, jump) that the object will have.

Once the class is created, you can make many objects from it — each one can have different values but the same structure.

```python
# example:

class Car:
    def __init__(self, brand, color):
        self.brand = brand
        self.color = color

    def drive(self):
        print(f"The {self.color} {self.brand} is driving.")

# Making objects from the class
car1 = Car("Toyota", "Red")
car2 = Car("BMW", "Black")

car1.drive()
car2.drive()
```

What is the difference between abstraction and encapsulation?
 - Abstraction means:

Showing only important things and hiding the unnecessary details.

Example:
When you drive a car, you only use the steering wheel, accelerator, and brakes.
You don’t worry about how the engine burns fuel or how the gears move inside — that part is hidden from you.
This is abstraction: you see only what you need to use.

In Python, we usually achieve abstraction by using abstract classes or interfaces (using the abc module).

 Encapsulation means:

Wrapping the data and methods together into one unit and keeping it safe from outside interference.

Example:
Imagine the car’s engine. It is protected inside the body of the car.
You cannot directly touch the engine while driving — it’s hidden and protected.
In coding, we do this by hiding data inside a class and restricting access to it using private or protected variables.

In Python, encapsulation is often done by prefixing variables with a single _ (protected) or double __ (private).

```python
# Example of Abstraction:

from abc import ABC, abstractmethod

# Abstract class
class Vehicle(ABC):
    
    @abstractmethod
    def start_engine(self):
        pass

# Concrete class
class Car(Vehicle):
    
    def start_engine(self):
        print("The car engine starts with a key.")

# Using the class
my_car = Car()
my_car.start_engine()
```

```python
# encapsulation Example:

class BankAccount:
    def __init__(self, balance):
        self.__balance = balance  # private variable

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount

    def get_balance(self):
        return self.__balance

# Using the class
account = BankAccount(1000)
account.deposit(500)
print(account.get_balance())  # Output: 1500

# account.__balance  # This will give an error! (because it’s private)
```

What are dunder methods in Python?
 - Dunder methods (short for "double underscore methods") are special methods in Python.
Their names start and end with two underscores — like __init__, __str__, __add__, etc.

They are also called magic methods because they let your objects behave like built-in Python types (such as behaving like numbers, strings, lists, etc.)

You don't directly call dunder methods yourself. Python automatically calls them when you do certain actions.

Explain the concept of inheritance in OOP?
-  Inheritance is when a child class gets properties and behaviors (methods) from a parent class.

It’s just like real life:

You inherit traits (like eye color or height) from your parents.

In coding, a new class (child) can reuse the code from an existing class (parent), and it can even add its own new things.

This makes coding faster, easier, and avoids repeating the same code again and again.

```python
# Example of Inheritance:

# Parent class
class Animal:
    def sound(self):
        print("Animals make sounds.")

# Child class
class Dog(Animal):
    def bark(self):
        print("Dog barks: Woof! Woof!")

# Create an object of Dog
dog = Dog()
dog.sound()  # Inherited from Animal class
dog.bark()   # Own method of Dog class
```

What is polymorphism in OOP?
- Polymorphism means:

"One thing behaving in many different ways."

In OOP, polymorphism allows the same function or method name to work differently depending on which object is calling it.

Just like in real life:

You can press the start button on a car, a laptop, or a washing machine.

Same button name (start), but different actions happen depending on the machine.

In Python, polymorphism lets us write code that works with different types of objects using the same method name.

```python
# Example of Polumorphism:

class Cat:
    def sound(self):
        print("Meow!")

class Dog:
    def sound(self):
        print("Woof!")

# Polymorphism in action
def make_animal_sound(animal):
    animal.sound()

# Create objects
cat = Cat()
dog = Dog()

make_animal_sound(cat)  # Output: Meow!
make_animal_sound(dog)  # Output: Woof!
```

How is encapsulation achieved in Python?
- Encapsulation is about keeping the data (variables) and the code (methods) that operates on the data together in one place, inside a class.
It also means protecting the data by making it private or protected, and providing controlled access to it through methods.

In Python, we achieve encapsulation using access modifiers:

Public: Variables or methods that can be accessed directly from outside the class (default in Python).

Protected: Variables or methods that should not be accessed directly, but can be inherited or used by subclasses (denoted by a single underscore _).

Private: Variables or methods that should not be accessed directly at all, even by subclasses (denoted by double underscores __).

```python
# example of Encapsulation:

class BankAccount:
    def __init__(self, balance):
        self.__balance = balance  # private variable
    
    # Public method to access private data
    def get_balance(self):
        return self.__balance
    
    # Public method to deposit money
    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount

# Create an object
account = BankAccount(1000)

# Access balance via public method (safe)
print(account.get_balance())  # Output: 1000

# You can't do this because __balance is private!
# print(account.__balance)  # This will cause an error
```

What are class and static methods in Python?
- A class method is a method that is bound to the class rather than the instance (object) of the class. It is defined using the @classmethod decorator and takes cls as its first parameter (which refers to the class itself, not the object).

Purpose: Class methods are used for operations that affect the class itself, not individual instances (objects). You can call them on the class directly or via an object.



A static method is a method that is not bound to either the instance or the class. It is defined using the @staticmethod decorator and doesn't take self or cls as its first parameter. Static methods behave like regular functions, but they belong to the class namespace.

Purpose: Static methods are used when you need to perform a task that is related to the class but doesn't need access to any instance or class-specific data.

```python
# Example of class method:

class Car:
    wheels = 4  # Class variable

    @classmethod
    def get_wheels(cls):
        return cls.wheels

# Using class method
print(Car.get_wheels())  # Output: 4
```

```python
# Example of static method:

class Math:
    @staticmethod
    def add(a, b):
        return a + b

# Using static method
print(Math.add(5, 3))  # Output: 8
```

What is method overloading in Python?
-  Method Overloading refers to the ability to define multiple methods with the same name but with different parameters (either number or type). However, Python does not support traditional method overloading like other programming languages (e.g., Java, C++).

In Python, only the last method with the same name gets used. So, if you define multiple methods with the same name, the one defined last will override the others. But don't worry — we can achieve similar functionality through default arguments or by using variable-length arguments (*args and **kwargs).

```python
# Example 1: Using Default Arguments:

class Calculator:
    def add(self, a, b=0):  # Default value for b
        return a + b

# Create object
calc = Calculator()

# Calling add with one or two arguments
print(calc.add(5))      # Output: 5 (uses default b=0)
print(calc.add(5, 3))   # Output: 8
```

```python
#Example 2: Using *args (Variable-Length Arguments):

class Printer:
    def print_info(self, *args):
        for arg in args:
            print(arg)

# Create object
printer = Printer()

# Calling with different numbers of arguments
printer.print_info("Hello")             # Output: Hello
printer.print_info("Hello", "World")    # Output: Hello \n World
printer.print_info("Hello", "World", 123) # Output: Hello \n World \n 123
```

What is method overriding in OOP?
- Method Overriding occurs when a child class provides a new implementation of a method that was already defined in the parent class.

The child class overrides the parent class’s method to give it specific behavior.

The method in the child class must have the same name, same parameters, and same return type as the method in the parent class.

Overriding allows the child class to customize or extend the behavior of the parent class without changing the parent class itself.

```python
# example of method overriding:

class Animal:
    def speak(self):
        print("Animal makes a sound")

class Dog(Animal):
    def speak(self):  # Overriding the parent method
        print("Dog barks")

# Creating objects
animal = Animal()
dog = Dog()

# Calling methods
animal.speak()  # Output: Animal makes a sound
dog.speak()     # Output: Dog barks
```

Why is polymorphism important in OOP?
-  Polymorphism is crucial because it provides flexibility and scalability in your code. Here's why it's important:

1. Code Reusability and Simplification:
Polymorphism allows we to reuse the same method name for different classes, making your code simpler and more maintainable. Instead of creating multiple methods with different names (e.g., dog_speak() and cat_speak()), we can use the same method name (speak()), and the behavior changes based on the object type. This reduces the need for repetitive code.

2. Flexibility and Extensibility:
With polymorphism, we can write generic code that works with any class that follows a certain interface (like having the speak() method), without worrying about the specific class. This is useful when you want to add new types of objects without changing the existing code.

For example, if we later add a Bird class, you don’t need to change the code that calls speak() — the behavior will automatically adjust to the new class.

3. Easier Maintenance and Updates:
Polymorphism makes it easy to maintain and update the code. If we need to change how a certain method works in one class, you don't need to search for all places where that method is used — it will automatically apply to all instances of that class and its subclasses.

4. Encapsulation of Behavior:
Polymorphism allows different classes to encapsulate their own behavior. Each class can define its own version of a method (e.g., speak()), and we don't need to worry about how they work internally.

```python
# example:

class Car:
    def start(self):
        print("Car engine starts.")

class Bike:
    def start(self):
        print("Bike engine starts.")

class Airplane:
    def start(self):
        print("Airplane engine starts.")

def start_vehicle(vehicle):
    vehicle.start()

# Using polymorphism
car = Car()
bike = Bike()
airplane = Airplane()

start_vehicle(car)       # Output: Car engine starts.
start_vehicle(bike)      # Output: Bike engine starts.
start_vehicle(airplane)  # Output: Airplane engine starts.
```

What is an abstract class in Python?
- Abstract Class:
An abstract class is a class that cannot be instantiated (you cannot create objects directly from it). It’s meant to serve as a blueprint for other classes. Abstract classes are used when you want to define a general structure and leave the implementation of certain methods to be done in child classes.

In Python, we use the abc module (Abstract Base Class) to define abstract classes and abstract methods.

1. Abstract Methods:
An abstract method is a method that is declared in the abstract class but does not have any implementation. It must be implemented in any child class that inherits from the abstract class.

```python
# Example of Abstract Class in Python:

from abc import ABC, abstractmethod

# Defining an abstract class
class Animal(ABC):
    @abstractmethod
    def speak(self):
        pass  # No implementation here, just a declaration

class Dog(Animal):
    def speak(self):
        print("Dog barks")

class Cat(Animal):
    def speak(self):
        print("Cat meows")

# Creating objects of Dog and Cat
dog = Dog()
cat = Cat()

dog.speak()  # Output: Dog barks
cat.speak()  # Output: Cat meows
```

What is the difference between a class variable and an instance variable?
- 1. Instance Variables:
Definition: Instance variables are specific to each object (instance) created from the class. They hold data that is unique to each instance of the class.

Where they are defined: Instance variables are usually defined inside the __init__() method of the class, using self.

Accessed: We can access instance variables using self.variable_name, and each instance of the class can have its own unique values for these variables.

Example: Think of a Car class. Each Car object can have a different color, model, and year. These would be instance variables.

2. Class Variables:
Definition: Class variables are shared across all instances of the class. They are not specific to any one object, but instead, they belong to the class itself.

Where they are defined: Class variables are defined inside the class, but outside of any methods.

Accessed: We can access class variables using the class name or via self, but they are the same for all objects unless modified.

Example: In the Car class, a class variable could represent the total number of cars created, since this value is the same for all instances.

```python
# example

class Car:
    total_cars = 0  # Class variable, shared by all instances
    
    def __init__(self, brand, model, color):
        self.brand = brand  # Instance variable
        self.model = model  # Instance variable
        self.color = color  # Instance variable
        Car.total_cars += 1  # Updating class variable whenever a new car is created

# Create instances of Car
car1 = Car("Toyota", "Corolla", "Red")
car2 = Car("Honda", "Civic", "Blue")

# Accessing instance variables
print(car1.brand)  # Output: Toyota
print(car2.model)  # Output: Civic

# Accessing class variable
print(Car.total_cars)  # Output: 2 (total number of cars created)
print(car1.total_cars)  # Output: 2 (can also be accessed via an instance)
```

What is multiple inheritance in Python?
- Multiple Inheritance occurs when a class inherits from more than one parent class. This means the child class can inherit attributes and methods from multiple classes.

Why is it useful?
It allows you to combine features from different classes and create a class with behavior and properties from more than one source. This can be useful when you need a class to have the characteristics of multiple existing classes without duplicating code.

```python
# Example of Multiple Inheritance:

class Animal:
    def speak(self):
        print("Animal makes a sound")

class Mammal:
    def has_hair(self):
        print("Mammals have hair")

class Dog(Animal, Mammal):  # Inheriting from both Animal and Mammal
    def bark(self):
        print("Dog barks")

# Creating object of Dog
dog = Dog()

# Accessing methods from both parent classes
dog.speak()     # Output: Animal makes a sound (inherited from Animal)
dog.has_hair()  # Output: Mammals have hair (inherited from Mammal)
dog.bark()      # Output: Dog barks (defined in Dog class)
```

Explain the purpose of ‘’__str__’ and ‘__repr__’ ‘ methods in Python?
1. __str__ Method:
Purpose: The __str__ method is used to define the informal or user-friendly string representation of an object. When you call str(object) or use print(object), Python calls this method to convert the object into a string.

Usage: This method is meant to return a string that is easy to read and understandable for the user.

2. __repr__ Method:
Purpose: The __repr__ method is used to define the formal string representation of an object. It is intended for developers and debugging purposes. When you call repr(object) or just type the object in the console (without print()), Python calls this method to convert the object into a string.

Usage: This method should return a string that ideally, when passed to eval(), would recreate the object.

```python
# Example:

class Car:
    def __init__(self, brand, model, year):
        self.brand = brand
        self.model = model
        self.year = year

    def __str__(self):
        return f"{self.year} {self.brand} {self.model}"  # User-friendly output

    def __repr__(self):
        return f"Car('{self.brand}', '{self.model}', {self.year})"  # Developer-friendly output

# Creating an object of Car
car = Car("Toyota", "Corolla", 2020)

# Calling str() and repr()
print(str(car))    # Output: 2020 Toyota Corolla (User-friendly)
print(repr(car))   # Output: Car('Toyota', 'Corolla', 2020) (Developer-friendly)

# Also, in the interactive console:
car  # Output: Car('Toyota', 'Corolla', 2020)
```

What is the significance of the ‘super()’ function in Python?
- super() Function:
The super() function is used to call methods from a parent class (also known as a superclass) within a child class (also called a subclass). It provides a way for a child class to invoke methods or access properties defined in its parent class without directly referencing the parent class by name.

Why is super() Important?
Inheritance: It helps in invoking the methods of a parent class when a class inherits from another class.

Avoids Redundancy: You don’t have to explicitly reference the parent class by its name, making the code more flexible and easier to maintain.

Works with Multiple Inheritance: In multiple inheritance scenarios, super() ensures that the method resolution order (MRO) is followed properly, allowing for the correct method from the appropriate class to be called.

How does super() work?
When super() is called, it looks up the method resolution order (MRO) and searches for the next class in the inheritance chain. It then calls the method from that class.

super() is commonly used within the __init__() method in child classes to initialize the parent class or call other methods.

```python
# Example of super():

class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        print(f"{self.name} makes a sound")

class Dog(Animal):
    def __init__(self, name, breed):
        super().__init__(name)  # Calling the __init__ method of the parent class
        self.breed = breed

    def speak(self):
        super().speak()  # Calling the speak() method of the parent class
        print(f"{self.name} barks")

# Creating an object of Dog
dog = Dog("Buddy", "Golden Retriever")

# Calling methods
dog.speak()
```

What is the significance of the __del__ method in Python?
- The __del__ method is called a destructor.
It is a special method in Python that automatically runs when an object is about to be destroyed, or deleted from memory.

Its job is to let you define cleanup actions, like closing files, releasing network connections, or printing a goodbye message — basically, anything you want to do just before the object disappears.

What is the difference between @staticmethod and @classmethod in Python?

But they serve different purposes and have different access levels:

@staticmethod — Like a regular function inside a class
Does not take self or cls as a first argument

Can’t access or modify class or instance attributes

Used when the method does something related to the class, but does not need access to the class or instance.

How does polymorphism work in Python with inheritance?
- Polymorphism simply means “many forms.”

In programming, it means that different classes can have methods with the same name, and Python can call the correct method depending on which object it is working with.

Even though the methods have the same name, the behavior depends on the object.

How does it connect with Inheritance?
When different classes inherit from a common parent class, and each one overrides a method (i.e., writes its own version of the method), polymorphism allows Python to treat objects from different classes the same way — but run their own method.

```python
# Example:

class Animal:
    def speak(self):
        print("Animal makes a sound")

class Dog(Animal):
    def speak(self):
        print("Dog barks")

class Cat(Animal):
    def speak(self):
        print("Cat meows")

# Polymorphism in action
def make_animal_speak(animal):
    animal.speak()

# Different objects, same method call
dog = Dog()
cat = Cat()

make_animal_speak(dog)  # Output: Dog barks
make_animal_speak(cat)  # Output: Cat meows
```

What is method chaining in Python OOP?
- Method Chaining is when you call multiple methods one after another on the same object in a single line of code.

Each method call returns the object itself (usually using return self), allowing you to "chain" another method to it immediately.

We can think of it like connecting Lego blocks one after another!

```python
# Example:

class Book:
    def __init__(self):
        self.pages = 0
        self.chapters = 0

    def add_pages(self, number):
        self.pages += number
        return self  # Returning the same object for chaining

    def add_chapters(self, number):
        self.chapters += number
        return self  # Returning the same object for chaining

    def display(self):
        print(f"Book has {self.pages} pages and {self.chapters} chapters")
        return self  # Returning the same object for further chaining if needed

# Using method chaining
book = Book()
book.add_pages(100).add_chapters(5).display()

# Output:
# Book has 100 pages and 5 chapters
```

What is the purpose of the __call__ method in Python?

__call__ is a special method in Python that lets you make an object behave like a function.

In simple words:
If a class has a __call__ method, we can "call" its objects like a function — with parentheses and arguments!

```python
# Example of Call()

class Adder:
    def __init__(self, number):
        self.number = number

    def __call__(self, x):
        return self.number + x

add_five = Adder(5)
print(add_five(10))  # Output: 15
```

```python
#2. Write a program to create an abstract class Shape with a method area(). Derive classes Circle and Rectangle 
# from it and implement the area() method in both.

from abc import ABC, abstractmethod

class shape (ABC):
    @abstractmethod
    def area (self):
        pass

class circle(shape) :
    def __init__(self,pi,r):
        self.pi= pi
        self.r = r

    def area (self):
        return (self.pi*(self.r**2))

class rectangle(shape):
    def __init__(self,length,width):
        self.length = length
        self.width = width

    def area(self):
        return self.length*self.width

```

```python
cir1 = circle(5,6)
cir1.area()
```

```python
rect1 = rectangle(8,9)
rect1.area()
```

```python
#3. Implement a multi-level inheritance scenario where a class Vehicle has an attribute type. Derive a class Car 
# and further derive a class ElectricCar that adds a battery attribute.


class Vehicle:
    def __init__(self, vehicle_type):
        self.vehicle_type = vehicle_type

    def display_type(self):
        print(f"Vehicle Type: {self.vehicle_type}")


class Car(Vehicle):
    def __init__(self, vehicle_type, brand):
        super().__init__(vehicle_type)
        self.brand = brand

    def display_brand(self):
        print(f"Car Brand: {self.brand}")


class ElectricCar(Car):
    def __init__(self, vehicle_type, brand, battery_capacity):
        super().__init__(vehicle_type, brand)
        self.battery_capacity = battery_capacity

    def display_battery(self):
        print(f"Battery Capacity: {self.battery_capacity} kWh")


ecar = ElectricCar("Four Wheeler", "Tesla", 75)
ecar.display_type()        
ecar.display_brand()       
ecar.display_battery()     
```

```python
#5. Write a program to demonstrate encapsulation by creating a class BankAccount with private attributes 
# balance and methods to deposit, withdraw, and check balance.



class BankAccount:
    def __init__(self, initial_balance):
        self.__balance = initial_balance  

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"Deposited: {amount}")
        else:
            print("Deposit amount must be positive.")

    def withdraw(self, amount):
        if amount <= self.__balance:
            self.__balance -= amount
            print(f"Withdrawn: {amount}")
        else:
            print("Insufficient balance.")

    def check_balance(self):
        print(f"Current Balance: {self.__balance}")


my_account = BankAccount(1000)


my_account.deposit(500)
my_account.withdraw(300)
my_account.check_balance()
```

```python
# 6. Demonstrate runtime polymorphism using a method play() in a base class Instrument. Derive classes Guitar 
# and Piano that implement their own version of play().


class Instrument:
    def play(self):
        print("Instrument is playing music.")


class Guitar(Instrument):
    def play(self):
        print("Guitar is playing chords.")


class Piano(Instrument):
    def play(self):
        print("Piano is playing melodies.")


inst1 = Guitar()
inst2 = Piano()

# Demonstrating runtime polymorphism
for instrument in [inst1, inst2]:
    instrument.play()
    
# Polymorphism in action:
# - We create objects of the Guitar and Piano classes.
# - Then, we loop through these objects and call the play() method.
# - At runtime, when we call play() on a Guitar object (inst1), Python executes
#   the play() method from the Guitar class, not the Instrument class.
# - Similarly, when we call play() on a Piano object (inst2), Python executes
#   the play() method from the Piano class, based on the actual object type.
```

```python
# 7. Create a class MathOperations with a class method add_numbers() to add two numbers and a static 
# method subtract_numbers() to subtract two numbers.

class MathOperations:
    # Class method to add two numbers
    @classmethod
    def add_numbers(cls, num1, num2):
        return num1 + num2

    # Static method to subtract two numbers
    @staticmethod
    def subtract_numbers(num1, num2):
        return num1 - num2

# Creating object of MathOperations to call the class method
math_ops = MathOperations()

# Using class method to add numbers
sum_result = math_ops.add_numbers(10, 5)
print(f"Sum: {sum_result}")  

# Using static method to subtract numbers
difference_result = MathOperations.subtract_numbers(10, 5)
print(f"Difference: {difference_result}")  
```

```python
# 8. Implement a class Person with a class method to count the total number of persons created.

class Person:
    
    total_persons = 0

    def __init__(self, name, age):
        self.name = name
        self.age = age
        Person.total_persons += 1  

    
    @classmethod
    def count_persons(cls):
        return cls.total_persons

# Creating person objects
person1 = Person("Alice", 25)
person2 = Person("Bob", 30)
person3 = Person("Charlie", 35)

# Using class method to count total persons
print(f"Total persons created: {Person.count_persons()}")  
```

```python
#9. Write a class Fraction with attributes numerator and denominator. Override the str method to display the 
# fraction as "numerator/denominator".

class Fraction:
    def __init__(self, numerator, denominator):
        self.numerator = numerator
        self.denominator = denominator

    # Overriding the __str__ method to display the fraction as "numerator/denominator"
    def __str__(self):
        return f"{self.numerator}/{self.denominator}"

# Creating a Fraction object
fraction = Fraction(3, 4)

# Printing the fraction using the __str__ method
print(fraction)
```

```python
# 10. Demonstrate operator overloading by creating a class Vector and overriding the add method to add two 
# vectors.

class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    # Overloading the '+' operator using __add__
    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)

    # String representation to print vector nicely
    def __str__(self):
        return f"({self.x}, {self.y})"

# Creating two vector objects
v1 = Vector(2, 4)
v2 = Vector(5, 7)

# Adding the two vectors using the overloaded '+' operator
v3 = v1 + v2

# Printing the result
print(f"Resultant Vector: {v3}")
```

```python
# 12. Implement a class Student with attributes name and grades. Create a method average_grade() to compute 
# the average of the grades.

class Student:
    def __init__(self, name, grades):
        self.name = name
        self.grades = grades  

    def average_grade(self):
        if len(self.grades) == 0:
            return 0
        return sum(self.grades) / len(self.grades)


student1 = Student("Aarav", [85, 90, 78, 92])


print(f"{student1.name}'s average grade is: {student1.average_grade():.2f}")
```

```python
# 14. Create a class Employee with a method calculate_salary() that computes the salary based on hours worked 
# and hourly rate. Create a derived class Manager that adds a bonus to the salary

class Employee:
    def __init__(self, name, hours_worked, hourly_rate):
        self.name = name
        self.hours_worked = hours_worked
        self.hourly_rate = hourly_rate

    def calculate_salary(self):
        return self.hours_worked * self.hourly_rate

class Manager(Employee):
    def __init__(self, name, hours_worked, hourly_rate, bonus):
        super().__init__(name, hours_worked, hourly_rate)
        self.bonus = bonus

    def calculate_salary(self):
        base_salary = super().calculate_salary()
        return base_salary + self.bonus

# Creating an Employee object
emp1 = Employee("Anita", 40, 20)
print(f"{emp1.name}'s Salary: ₹{emp1.calculate_salary()}")

# Creating a Manager object
mgr1 = Manager("Ravi", 40, 30, 5000)
print(f"{mgr1.name}'s Salary (with bonus): ₹{mgr1.calculate_salary()}")
```

```python
#15. Create a class Product with attributes name, price, and quantity. Implement a method total_price() that 
# calculates the total price of the product.

class Product:
    def __init__(self, name, price, quantity):
        self.name = name
        self.price = price
        self.quantity = quantity

    def total_price(self):
        return self.price * self.quantity

# Creating a product object
product1 = Product("Notebook", 50, 4)

# Calculating and printing the total price
print(f"Total price for {product1.quantity} {product1.name}s: ₹{product1.total_price()}")
```

```python
# 16. Create a class Animal with an abstract method sound(). Create two derived classes Cow and Sheep that 
# implement the sound() method.

from abc import ABC, abstractmethod


class Animal(ABC):
    @abstractmethod
    def sound(self):
        pass


class Cow(Animal):
    def sound(self):
        return "Moo"


class Sheep(Animal):
    def sound(self):
        return "Baa"


cow = Cow()
sheep = Sheep()

print("Cow says:", cow.sound())
print("Sheep says:", sheep.sound())
```

```python
#17. Create a class Book with attributes title, author, and year_published. Add a method get_book_info() that 
# returns a formatted string with the book's details.

class Book:
    def __init__(self, title, author, year_published):
        self.title = title
        self.author = author
        self.year_published = year_published

    def get_book_info(self):
        return f"'{self.title}' by {self.author} (Published in {self.year_published})"

# Creating a book object
book1 = Book("Marcus Aurelius Meditations", "George Long", 1862)

# Printing book information
print(book1.get_book_info())
```
